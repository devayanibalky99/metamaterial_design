### 1.0 Dataset Generation

```python
import pandas as pd
```

```python
def combinations(original_list):
    # The number of subsets is 2^n
    num_subsets = 2 ** len(original_list)

    # Create an empty list to hold all the subsets
    subsets = []

    # Iterate over all possible subsets
    for subset_index in range(num_subsets):
        # Create an empty list to hold the current subset
        subset = []
        # Iterate over all elements in the original list
        for index in range(len(original_list)):
            # Check if index bit is set in subset_index
            if (subset_index & (1 << index)) != 0:
                # If the bit is set, add the element at this index to the current subset
                subset.append(original_list[index])
        # Add the current subset to the list of all subsets
        subsets.append(subset)
        
    return subsets
```

```python
# Using the function to print generated combinations
lst = [12, 14, 15, 23, 24, 25, 26, 35, 36, 45, 47, 48, 56, 57, 58, 59, 68, 69, 78, 89]
all_combinations = pd.DataFrame (combinations(lst))
all_combinations.head (15)
```

```python
len(all_combinations.index)
```

```python
data_types = all_combinations.dtypes

print(data_types)
```

### 2.0 Data Cleaning

```python
#remove first row 
all_combinations_clean = all_combinations.tail(-1)
all_combinations_clean.head(5)
```

```python
#rename columns 
all_combinations_clean.columns = ['VC1', 'VC2', 'VC3', 'VC4', 'VC5', 'VC6', 'VC7', 'VC8', 'VC9', 'VC10',
                            'VC11', 'VC12', 'VC13', 'VC14', 'VC15', 'VC16', 'VC17', 'VC18', 'VC19', 'VC20']
all_combinations_clean.head(5)
```

```python
all_combinations_clean.isnull().sum()
```

```python
#drop columns without at least 4 VC
all_combinations_clean=all_combinations_clean.dropna(subset=["VC4"])
len(all_combinations_clean.index)
```

```python
#replace Nan with 0 for readability
all_combinations_clean.fillna(0, inplace=True)
all_combinations_clean.head(5)
```

```python
#change datatype 
all_combinations_clean = all_combinations_clean.astype(int)
all_combinations_clean.head(5)
```

```python
data_types = all_combinations_clean.dtypes
print(data_types)
```

```python
#save cleaned df as csv
all_combinations_clean.to_csv("all_combinations.csv", index=False)
```

```python
#save selection from cleaned df as csv
all_combinations_selection = all_combinations_clean.sample(n = 1000, random_state = 145, replace = False)
all_combinations_selection.to_csv("all_combinations_selection.csv", index=False)
```
